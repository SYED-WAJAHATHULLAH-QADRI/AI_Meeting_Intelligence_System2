```python
# ============================================================
# MEETING DATA UPLOAD FUNCTION
# ============================================================

from google.colab import files
import shutil
from pathlib import Path

def upload_meeting_files(meeting_id):
    """
    Upload one audio file and one human/reference transcript
    for a specified meeting ID.
    """

    if meeting_id not in dataset_manifest["meeting_id"].values:
        print(f"ERROR: {meeting_id} is not present in the dataset manifest.")
        return

    print(f"Preparing files for {meeting_id}")
    print("\n1. Upload the AUDIO file first.")
    print("Accepted examples: .wav, .mp3, .m4a, .mp4")

    audio_upload = files.upload()

    if len(audio_upload) != 1:
        print("ERROR: Please upload exactly one audio file.")
        return

    audio_name = list(audio_upload.keys())[0]

    audio_destination = (
        DIRS["audio"] /
        f"{meeting_id}_{audio_name}"
    )

    with open(audio_destination, "wb") as f:
        f.write(audio_upload[audio_name])

    print("\nAudio saved:")
    print(audio_destination)

    print("\n2. Now upload the HUMAN/REFERENCE transcript.")
    print("Use a plain .txt transcript if possible.")

    transcript_upload = files.upload()

    if len(transcript_upload) != 1:
        print("ERROR: Please upload exactly one transcript file.")
        return

    transcript_name = list(transcript_upload.keys())[0]

    transcript_destination = (
        DIRS["reference_transcripts"] /
        f"{meeting_id}_{transcript_name}"
    )

    with open(transcript_destination, "wb") as f:
        f.write(transcript_upload[transcript_name])

    # Update manifest
    row_index = dataset_manifest.index[
        dataset_manifest["meeting_id"] == meeting_id
    ][0]

    dataset_manifest.loc[
        row_index,
        "audio_filename"
    ] = audio_destination.name

    dataset_manifest.loc[
        row_index,
        "reference_transcript_filename"
    ] = transcript_destination.name

    dataset_manifest.loc[
        row_index,
        "audio_available"
    ] = True

    dataset_manifest.loc[
        row_index,
        "reference_transcript_available"
    ] = True

    dataset_manifest.loc[
        row_index,
        "ready_for_asr_evaluation"
    ] = True

    dataset_manifest.to_csv(
        DIRS["data"] / "dataset_manifest.csv",
        index=False
    )

    print("\nMeeting files registered successfully.")
    print("Meeting ID:", meeting_id)
    print("Audio:", audio_destination.name)
    print("Transcript:", transcript_destination.name)
    print("Ready for ASR evaluation: True")


print("Meeting upload function created successfully.")
print("Next meeting to prepare: M01")
```

```python
# ============================================================
# CREATE CLEAN ASR REFERENCE FOR M01
# Remove document metadata and speaker labels for fair WER/CER
# ============================================================

import re

meeting_id = "M01"

# Start from the reference_text loaded in Step 16
lines = reference_text.splitlines()

clean_lines = []

metadata_prefixes = (
    "Meeting Transcript",
    "Meeting Topic:",
    "Duration:",
    "Participants:"
)

for line in lines:

    line = line.strip()

    if not line:
        continue

    # Remove document metadata/header lines
    if line.startswith(metadata_prefixes):
        continue

    # Remove speaker labels such as:
    # Sarah:
    # James:
    # Michael:
    # Aisha:
    line = re.sub(
        r"^[A-Za-z][A-Za-z .'-]{0,40}:\s*",
        "",
        line
    )

    if line:
        clean_lines.append(line)

clean_reference_text = " ".join(clean_lines)

# Normalize whitespace
clean_reference_text = re.sub(
    r"\s+",
    " ",
    clean_reference_text
).strip()

# Save clean ASR reference
clean_reference_file = (
    DIRS["reference_transcripts"] /
    "M01_ASR_reference_clean.txt"
)

clean_reference_file.write_text(
    clean_reference_text,
    encoding="utf-8"
)

print("Clean ASR reference created successfully.")
print("Original reference word count:", len(reference_text.split()))
print(
    "Clean ASR reference word count:",
    len(clean_reference_text.split())
)

print("\nFirst 1000 characters of CLEAN reference:\n")
print(clean_reference_text[:1000])

print("\nSaved file:")
print(clean_reference_file)
```

```python
# ============================================================
# M01 ASR ACCURACY: WER AND CER
# ============================================================

import re
import unicodedata
from jiwer import wer, cer


def normalize_asr_text(text):
    """
    Normalize reference and Whisper transcript equally
    before calculating WER/CER.
    """

    # Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # Normalize apostrophes
    text = text.replace("’", "'")
    text = text.replace("‘", "'")

    # Lowercase
    text = text.lower()

    # Normalize percentage symbol
    text = text.replace("%", " percent ")

    # Remove punctuation but preserve letters/numbers/apostrophes
    text = re.sub(
        r"[^a-z0-9'\s]",
        " ",
        text
    )

    # Normalize whitespace
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text


# IMPORTANT:
# Use FULL reference because the audio itself contains
# headings and speaker names.
reference_normalized = normalize_asr_text(reference_text)
whisper_normalized = normalize_asr_text(whisper_transcript)

m01_wer = wer(
    reference_normalized,
    whisper_normalized
)

m01_cer = cer(
    reference_normalized,
    whisper_normalized
)

print("M01 ASR EVALUATION")
print("=" * 50)

print(
    "Reference words:",
    len(reference_normalized.split())
)

print(
    "Whisper words:",
    len(whisper_normalized.split())
)

print(
    "Word Error Rate (WER):",
    round(m01_wer, 4)
)

print(
    "WER percentage:",
    round(m01_wer * 100, 2),
    "%"
)

print(
    "Character Error Rate (CER):",
    round(m01_cer, 4)
)

print(
    "CER percentage:",
    round(m01_cer * 100, 2),
    "%"
)

print(
    "Approximate Word Accuracy:",
    round((1 - m01_wer) * 100, 2),
    "%"
)
```

```python
# ============================================================
# DEFINE GENERIC AND STRUCTURED PROMPTS
# ============================================================

GENERIC_PROMPT = """
Read the following meeting transcript and produce useful meeting notes.

Identify:
- a short meeting summary
- key topics
- decisions
- action items
- owners
- deadlines
- unresolved issues

Return the result clearly and concisely.

MEETING TRANSCRIPT:
{transcript}
"""


STRUCTURED_PROMPT = """
You are analysing a meeting transcript.

Your task is to extract only information that is supported by the transcript.

Follow these rules strictly:

1. A DECISION must only be recorded if participants clearly agreed,
   confirmed, approved, selected, or decided something.

2. Do NOT treat:
   - suggestions
   - possibilities
   - opinions
   - rejected proposals
   - unanswered questions
   as confirmed decisions.

3. An ACTION ITEM must describe a specific task that somebody is expected
   to perform.

4. If an action owner is not explicitly stated, return null.

5. If a deadline is not explicitly stated, return null.

6. Do not invent missing names, dates, deadlines, tasks, decisions,
   responsibilities, or facts.

7. For every decision and action item, provide supporting evidence from
   the transcript.

8. Separate unresolved issues from confirmed decisions.

9. If information is ambiguous, mark it as ambiguous rather than guessing.

Return the following categories:

- meeting_title
- summary
- key_topics
- decisions
- action_items
- unresolved_issues
- ambiguities

Each decision should contain:
- decision
- evidence

Each action item should contain:
- action
- owner
- deadline
- evidence

MEETING TRANSCRIPT:
{transcript}
"""


print("Generic prompt created successfully.")
print("Structured prompt created successfully.")

print("\nGeneric prompt length:")
print(len(GENERIC_PROMPT), "characters")

print("\nStructured prompt length:")
print(len(STRUCTURED_PROMPT), "characters")

print("\nPrompt comparison ready for controlled evaluation.")
```

```python
# ============================================================
# COMMON OUTPUT SCHEMA FOR BOTH PROMPT CONDITIONS
# ============================================================

from pydantic import BaseModel, Field
from typing import List, Optional


class Decision(BaseModel):
    decision: str
    evidence: Optional[str] = None


class ActionItem(BaseModel):
    action: str
    owner: Optional[str] = None
    deadline: Optional[str] = None
    evidence: Optional[str] = None


class MeetingIntelligence(BaseModel):
    meeting_title: str
    summary: str
    key_topics: List[str]

    decisions: List[Decision] = Field(
        default_factory=list
    )

    action_items: List[ActionItem] = Field(
        default_factory=list
    )

    unresolved_issues: List[str] = Field(
        default_factory=list
    )

    ambiguities: List[str] = Field(
        default_factory=list
    )


print("Common MeetingIntelligence schema created successfully.")

print("\nRequired comparison fields:")
print("- meeting_title")
print("- summary")
print("- key_topics")
print("- decisions")
print("- action_items")
print("- unresolved_issues")
print("- ambiguities")

print("\nDecision fields:")
print("- decision")
print("- evidence")

print("\nAction-item fields:")
print("- action")
print("- owner")
print("- deadline")
print("- evidence")

print("\nSchema ready for generic vs structured comparison.")
```

```python
# ============================================================
# CONTROLLED GEMINI EXPERIMENT FUNCTION
# Generic vs Structured Prompt
# ============================================================

import json
import time
from datetime import datetime

GEMINI_TEMPERATURE = 0.2


def run_gemini_experiment(
    transcript,
    prompt_type,
    meeting_id,
    run_number=1
):
    """
    Run one controlled Gemini extraction experiment.

    Conditions kept constant:
    - Same Gemini model
    - Same temperature
    - Same transcript
    - Same output schema

    Only the prompt instructions differ.
    """

    if prompt_type not in ["generic", "structured"]:
        raise ValueError(
            "prompt_type must be 'generic' or 'structured'"
        )

    if prompt_type == "generic":
        prompt = GENERIC_PROMPT.format(
            transcript=transcript
        )
    else:
        prompt = STRUCTURED_PROMPT.format(
            transcript=transcript
        )

    print(
        f"Running {prompt_type} prompt | "
        f"Meeting: {meeting_id} | "
        f"Run: {run_number}"
    )

    start_time = time.perf_counter()

    try:

        response = gemini_client.models.generate_content(
            model=GEMINI_MODEL,
            contents=prompt,
            config={
                "temperature": GEMINI_TEMPERATURE,

                "response_format": {
                    "text": {
                        "mime_type": "application/json",
                        "schema": MeetingIntelligence.model_json_schema()
                    }
                }
            }
        )

        latency = time.perf_counter() - start_time

        # Validate returned JSON using Pydantic
        parsed_output = MeetingIntelligence.model_validate_json(
            response.text
        )

        experiment_record = {
            "meeting_id": meeting_id,
            "prompt_type": prompt_type,
            "run_number": run_number,
            "model": GEMINI_MODEL,
            "temperature": GEMINI_TEMPERATURE,
            "status": "SUCCESS",
            "latency_seconds": round(latency, 3),
            "timestamp": datetime.now().isoformat(),
            "error": ""
        }

        print("Status: SUCCESS")
        print(
            "Latency:",
            round(latency, 3),
            "seconds"
        )

        return parsed_output, experiment_record

    except Exception as e:

        latency = time.perf_counter() - start_time

        experiment_record = {
            "meeting_id": meeting_id,
            "prompt_type": prompt_type,
            "run_number": run_number,
            "model": GEMINI_MODEL,
            "temperature": GEMINI_TEMPERATURE,
            "status": "FAILED",
            "latency_seconds": round(latency, 3),
            "timestamp": datetime.now().isoformat(),
            "error": str(e)
        }

        print("Status: FAILED")
        print("Error:", str(e))

        return None, experiment_record


print("Controlled Gemini experiment function created successfully.")
print("Model:", GEMINI_MODEL)
print("Temperature:", GEMINI_TEMPERATURE)
print("Prompt conditions: generic vs structured")
print("Common output schema: MeetingIntelligence")
print("Function has NOT been executed yet.")
```

```python
# ============================================================
# HUMAN ANNOTATOR AGREEMENT AND CONSENSUS FUNCTION
# ============================================================

from difflib import SequenceMatcher
import pandas as pd


def text_similarity(text1, text2):
    """
    Calculate simple text similarity between two annotations.
    Returns a score between 0 and 1.
    """

    text1 = str(text1).lower().strip()
    text2 = str(text2).lower().strip()

    return SequenceMatcher(
        None,
        text1,
        text2
    ).ratio()


def compare_annotators(
    annotator1_data,
    annotator2_data,
    similarity_threshold=0.70
):
    """
    Compare A1 and A2 decisions/action items.

    Items with similarity >= threshold are considered
    provisional matches for later human consensus review.
    """

    comparison_rows = []

    for item_type in ["decision", "action_item"]:

        a1_items = annotator1_data[
            annotator1_data["item_type"] == item_type
        ]

        a2_items = annotator2_data[
            annotator2_data["item_type"] == item_type
        ]

        matched_a2 = set()

        for _, row1 in a1_items.iterrows():

            best_match = None
            best_similarity = 0
            best_index = None

            for index2, row2 in a2_items.iterrows():

                if index2 in matched_a2:
                    continue

                similarity = text_similarity(
                    row1["item_text"],
                    row2["item_text"]
                )

                if similarity > best_similarity:
                    best_similarity = similarity
                    best_match = row2
                    best_index = index2

            if (
                best_match is not None
                and best_similarity >= similarity_threshold
            ):

                matched_a2.add(best_index)

                comparison_rows.append({
                    "item_type": item_type,
                    "annotator1_text": row1["item_text"],
                    "annotator2_text": best_match["item_text"],
                    "text_similarity": round(best_similarity, 3),
                    "provisional_agreement": True,
                    "consensus_text": "",
                    "consensus_owner": "",
                    "consensus_deadline": "",
                    "consensus_evidence": ""
                })

            else:

                comparison_rows.append({
                    "item_type": item_type,
                    "annotator1_text": row1["item_text"],
                    "annotator2_text": "",
                    "text_similarity": round(best_similarity, 3),
                    "provisional_agreement": False,
                    "consensus_text": "",
                    "consensus_owner": "",
                    "consensus_deadline": "",
                    "consensus_evidence": ""
                })

        # Add unmatched A2 items
        for index2, row2 in a2_items.iterrows():

            if index2 not in matched_a2:

                comparison_rows.append({
                    "item_type": item_type,
                    "annotator1_text": "",
                    "annotator2_text": row2["item_text"],
                    "text_similarity": 0.0,
                    "provisional_agreement": False,
                    "consensus_text": "",
                    "consensus_owner": "",
                    "consensus_deadline": "",
                    "consensus_evidence": ""
                })

    comparison_df = pd.DataFrame(comparison_rows)

    if len(comparison_df) > 0:

        agreement_rate = (
            comparison_df["provisional_agreement"].mean()
        )

        print("Annotator comparison completed.")
        print(
            "Provisional agreement rate:",
            round(agreement_rate * 100, 2),
            "%"
        )

    else:
        print("No annotations available for comparison.")

    return comparison_df


print("Annotator comparison function created successfully.")
print("Similarity threshold: 0.70")
print("A1 and A2 files are NOT compared yet.")
print("Consensus must still be confirmed by a human reviewer.")
```

```python
# ============================================================
# EXTRACTION ACCURACY METRIC CALCULATOR
# Precision, Recall, F1, Hallucination,
# Owner Accuracy and Deadline Accuracy
# ============================================================

import pandas as pd


def calculate_extraction_metrics(
    meeting_id,
    prompt_type,
    run_number,
    item_type,
    true_positives,
    false_positives,
    false_negatives,
    owner_correct=0,
    owner_evaluable=0,
    deadline_correct=0,
    deadline_evaluable=0
):
    """
    Calculate extraction performance against the
    human-consensus gold standard.

    TP = AI item correctly matches gold standard
    FP = AI generated unsupported/incorrect item
    FN = Gold-standard item missed by AI
    """

    # --------------------------------------------------------
    # Precision
    # --------------------------------------------------------

    if (true_positives + false_positives) > 0:
        precision = (
            true_positives /
            (true_positives + false_positives)
        )
    else:
        precision = 0.0

    # --------------------------------------------------------
    # Recall
    # --------------------------------------------------------

    if (true_positives + false_negatives) > 0:
        recall = (
            true_positives /
            (true_positives + false_negatives)
        )
    else:
        recall = 0.0

    # --------------------------------------------------------
    # F1 Score
    # --------------------------------------------------------

    if (precision + recall) > 0:
        f1 = (
            2 * precision * recall /
            (precision + recall)
        )
    else:
        f1 = 0.0

    # --------------------------------------------------------
    # Hallucination Rate
    # FP / all AI-generated items
    # --------------------------------------------------------

    generated_items = (
        true_positives +
        false_positives
    )

    if generated_items > 0:
        hallucination_rate = (
            false_positives /
            generated_items
        )
    else:
        hallucination_rate = 0.0

    # --------------------------------------------------------
    # Owner Accuracy
    # --------------------------------------------------------

    if owner_evaluable > 0:
        owner_accuracy = (
            owner_correct /
            owner_evaluable
        )
    else:
        owner_accuracy = None

    # --------------------------------------------------------
    # Deadline Accuracy
    # --------------------------------------------------------

    if deadline_evaluable > 0:
        deadline_accuracy = (
            deadline_correct /
            deadline_evaluable
        )
    else:
        deadline_accuracy = None

    result = {
        "meeting_id": meeting_id,
        "prompt_type": prompt_type,
        "run_number": run_number,
        "item_type": item_type,

        "true_positives": true_positives,
        "false_positives": false_positives,
        "false_negatives": false_negatives,

        "precision": round(precision, 4),
        "recall": round(recall, 4),
        "f1_score": round(f1, 4),

        "hallucination_rate": round(
            hallucination_rate,
            4
        ),

        "owner_correct": owner_correct,
        "owner_evaluable": owner_evaluable,

        "owner_accuracy": (
            round(owner_accuracy, 4)
            if owner_accuracy is not None
            else None
        ),

        "deadline_correct": deadline_correct,
        "deadline_evaluable": deadline_evaluable,

        "deadline_accuracy": (
            round(deadline_accuracy, 4)
            if deadline_accuracy is not None
            else None
        )
    }

    return result


print("Extraction metric calculator created successfully.")

print("\nMetrics supported:")
print("- Precision")
print("- Recall")
print("- F1 score")
print("- Hallucination rate")
print("- Owner accuracy")
print("- Deadline accuracy")

print(
    "\nIMPORTANT: TP, FP and FN must be determined "
    "against the human-consensus gold standard."
)

print(
    "No experimental metric values have been calculated yet."
)
```

```python
# ============================================================
# SAVE GEMINI EXPERIMENT OUTPUTS AND RUN METADATA
# ============================================================

import json
import pandas as pd
from pathlib import Path


def save_gemini_experiment_result(
    parsed_output,
    experiment_record
):
    """
    Save one Gemini experimental run.

    Saves:
    1. Run metadata to CSV
    2. Successful structured output to JSON
    3. Failed-run information without inventing output
    """

    meeting_id = experiment_record["meeting_id"]
    prompt_type = experiment_record["prompt_type"]
    run_number = experiment_record["run_number"]

    # --------------------------------------------------------
    # Select output directory
    # --------------------------------------------------------

    if prompt_type == "generic":
        output_directory = DIRS["generic_outputs"]

    elif prompt_type == "structured":
        output_directory = DIRS["structured_outputs"]

    else:
        raise ValueError(
            "Unknown prompt type."
        )

    # --------------------------------------------------------
    # Save experiment metadata
    # --------------------------------------------------------

    metadata_file = (
        DIRS["extraction_results"] /
        "gemini_experiment_runs.csv"
    )

    new_record_df = pd.DataFrame(
        [experiment_record]
    )

    if metadata_file.exists():

        existing_df = pd.read_csv(
            metadata_file
        )

        # Prevent accidental duplicate record
        existing_df = existing_df[
            ~(
                (existing_df["meeting_id"] == meeting_id)
                &
                (existing_df["prompt_type"] == prompt_type)
                &
                (existing_df["run_number"] == run_number)
            )
        ]

        metadata_df = pd.concat(
            [
                existing_df,
                new_record_df
            ],
            ignore_index=True
        )

    else:

        metadata_df = new_record_df


    metadata_df.to_csv(
        metadata_file,
        index=False
    )

    # --------------------------------------------------------
    # Save successful structured output
    # --------------------------------------------------------

    if (
        experiment_record["status"] == "SUCCESS"
        and parsed_output is not None
    ):

        output_file = (
            output_directory /
            f"{meeting_id}_{prompt_type}_run{run_number}.json"
        )

        if hasattr(parsed_output, "model_dump"):
            output_data = parsed_output.model_dump()
        else:
            output_data = parsed_output

        with open(
            output_file,
            "w",
            encoding="utf-8"
        ) as f:

            json.dump(
                output_data,
                f,
                indent=2,
                ensure_ascii=False
            )

        print("Gemini output saved successfully.")
        print("Output file:")
        print(output_file)

    else:

        print(
            "Run was not successful, so no AI output JSON was created."
        )

        output_file = None

    print("\nRun metadata saved:")
    print(metadata_file)

    return output_file


print("Gemini experiment result saver created successfully.")

print("\nIt will record:")
print("- Meeting ID")
print("- Prompt type")
print("- Run number")
print("- Gemini model")
print("- Temperature")
print("- Success/failure status")
print("- Processing latency")
print("- Timestamp")
print("- HTTP 403 status")
print("- HTTP 429 quota status")
print("- Error message")
print("- Successful structured JSON output")

print("\nNo Gemini request has been made in this cell.")
```

```python
# ============================================================
# CONTENT-LEVEL REPEATABILITY ANALYSIS
# Compare extracted decisions and action items across runs
# ============================================================

import json
import re
import pandas as pd
import numpy as np
from rapidfuzz.fuzz import token_set_ratio


# ------------------------------------------------------------
# Text normalization
# ------------------------------------------------------------

def normalize_repeatability_text(text):

    if text is None:
        return ""

    text = str(text).lower().strip()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


# ------------------------------------------------------------
# Load one saved Gemini run
# ------------------------------------------------------------

def load_saved_run(
    meeting_id,
    prompt_type,
    run_number
):

    if prompt_type == "generic":
        directory = DIRS["generic_outputs"]

    elif prompt_type == "structured":
        directory = DIRS["structured_outputs"]

    else:
        raise ValueError(
            "prompt_type must be generic or structured"
        )

    file_path = (
        directory /
        f"{meeting_id}_{prompt_type}_run{run_number}.json"
    )

    with open(
        file_path,
        "r",
        encoding="utf-8"
    ) as f:
        return json.load(f)


# ------------------------------------------------------------
# Extract item text
# ------------------------------------------------------------

def get_item_texts(
    output_dict,
    item_type
):

    if item_type == "decisions":

        return [
            item["decision"]
            for item in output_dict.get(
                "decisions",
                []
            )
        ]

    elif item_type == "action_items":

        return [
            item["action"]
            for item in output_dict.get(
                "action_items",
                []
            )
        ]

    else:
        raise ValueError(
            "Unknown item type"
        )


# ------------------------------------------------------------
# One-to-one matching between two runs
# ------------------------------------------------------------

def compare_run_items(
    items_a,
    items_b,
    similarity_threshold=70
):

    candidates = []

    for i, text_a in enumerate(items_a):

        for j, text_b in enumerate(items_b):

            similarity = token_set_ratio(
                normalize_repeatability_text(
                    text_a
                ),
                normalize_repeatability_text(
                    text_b
                )
            )

            candidates.append({
                "index_a": i,
                "index_b": j,
                "text_a": text_a,
                "text_b": text_b,
                "similarity": similarity
            })


    # Highest similarity first
    candidates = sorted(
        candidates,
        key=lambda x: x["similarity"],
        reverse=True
    )


    used_a = set()
    used_b = set()
    matches = []


    for candidate in candidates:

        if (
            candidate["index_a"]
            not in used_a
            and
            candidate["index_b"]
            not in used_b
            and
            candidate["similarity"]
            >= similarity_threshold
        ):

            matches.append(candidate)

            used_a.add(
                candidate["index_a"]
            )

            used_b.add(
                candidate["index_b"]
            )


    matched_count = len(matches)

    # Stability precision/recall-like overlap
    # based on two generated sets
    denominator = (
        len(items_a) +
        len(items_b)
    )

    if denominator > 0:

        overlap_f1 = (
            2 * matched_count /
            denominator
        )

    else:
        overlap_f1 = 1.0


    if matched_count > 0:

        mean_similarity = np.mean([
            match["similarity"]
            for match in matches
        ])

    else:
        mean_similarity = 0.0


    return {
        "items_run_a": len(items_a),
        "items_run_b": len(items_b),
        "matched_items": matched_count,
        "overlap_f1": round(
            overlap_f1,
            4
        ),
        "mean_matched_similarity": round(
            mean_similarity / 100,
            4
        )
    }


# ============================================================
# PAIRWISE COMPARISON
# ============================================================

pairwise_rows = []

run_pairs = [
    (1, 2),
    (1, 3),
    (2, 3)
]


for prompt_type in [
    "generic",
    "structured"
]:

    outputs = {
        run: load_saved_run(
            "M01",
            prompt_type,
            run
        )
        for run in [1, 2, 3]
    }


    for item_type in [
        "decisions",
        "action_items"
    ]:

        for run_a, run_b in run_pairs:

            items_a = get_item_texts(
                outputs[run_a],
                item_type
            )

            items_b = get_item_texts(
                outputs[run_b],
                item_type
            )

            comparison = compare_run_items(
                items_a,
                items_b,
                similarity_threshold=70
            )

            pairwise_rows.append({
                "meeting_id": "M01",
                "prompt_type": prompt_type,
                "item_type": item_type,
                "run_a": run_a,
                "run_b": run_b,
                **comparison
            })


content_repeatability_df = pd.DataFrame(
    pairwise_rows
)


# ------------------------------------------------------------
# Summary by prompt and item type
# ------------------------------------------------------------

content_stability_summary = (
    content_repeatability_df
    .groupby([
        "prompt_type",
        "item_type"
    ])
    .agg(
        mean_overlap_f1=(
            "overlap_f1",
            "mean"
        ),
        sd_overlap_f1=(
            "overlap_f1",
            "std"
        ),
        mean_text_similarity=(
            "mean_matched_similarity",
            "mean"
        ),
        minimum_overlap_f1=(
            "overlap_f1",
            "min"
        )
    )
    .round(4)
    .reset_index()
)


# ------------------------------------------------------------
# Save results
# ------------------------------------------------------------

pairwise_file = (
    DIRS["repeatability"] /
    "M01_content_repeatability_pairwise.csv"
)

summary_file = (
    DIRS["repeatability"] /
    "M01_content_stability_summary.csv"
)


content_repeatability_df.to_csv(
    pairwise_file,
    index=False
)

content_stability_summary.to_csv(
    summary_file,
    index=False
)


print(
    "Content-level repeatability analysis completed successfully."
)

print(
    "\nSimilarity threshold: 70/100"
)

print(
    "\nPAIRWISE RUN COMPARISON:"
)

display(
    content_repeatability_df
)


print(
    "\nCONTENT STABILITY SUMMARY:"
)

display(
    content_stability_summary
)


print("\nSaved files:")
print(pairwise_file)
print(summary_file)


print(
    "\nINTERPRETATION WARNING:"
)

print(
    "Overlap F1 measures consistency between repeated AI runs."
)

print(
    "It does NOT measure correctness against the human gold standard."
)

print(
    "A system can be consistently wrong, so accuracy will be evaluated separately."
)
```

```python
# ============================================================
# BLINDED HUMAN EVALUATION PACKAGE
# M01 - GENERIC vs STRUCTURED - RUN 1
# ============================================================

import json
import random
from pathlib import Path
from openpyxl import Workbook
from openpyxl.styles import Font, Alignment


# ------------------------------------------------------------
# Load Run 1 outputs
# ------------------------------------------------------------

generic_run1_path = (
    DIRS["generic_outputs"] /
    "M01_generic_run1.json"
)

structured_run1_path = (
    DIRS["structured_outputs"] /
    "M01_structured_run1.json"
)

with open(
    generic_run1_path,
    "r",
    encoding="utf-8"
) as f:
    generic_run1_data = json.load(f)

with open(
    structured_run1_path,
    "r",
    encoding="utf-8"
) as f:
    structured_run1_data = json.load(f)


# ------------------------------------------------------------
# Convert structured JSON to readable text
# ------------------------------------------------------------

def format_meeting_output(data):

    sections = []

    sections.append(
        f"MEETING TITLE\n{data.get('meeting_title', '')}"
    )

    sections.append(
        f"SUMMARY\n{data.get('summary', '')}"
    )

    topics = "\n".join(
        f"- {x}"
        for x in data.get("key_topics", [])
    )

    sections.append(
        f"KEY TOPICS\n{topics}"
    )

    decisions = []

    for i, item in enumerate(
        data.get("decisions", []),
        start=1
    ):
        decisions.append(
            f"{i}. {item.get('decision', '')}\n"
            f"Evidence: {item.get('evidence', '')}"
        )

    sections.append(
        "DECISIONS\n" +
        "\n\n".join(decisions)
    )

    actions = []

    for i, item in enumerate(
        data.get("action_items", []),
        start=1
    ):
        actions.append(
            f"{i}. {item.get('action', '')}\n"
            f"Owner: {item.get('owner', '')}\n"
            f"Deadline: {item.get('deadline', '')}\n"
            f"Evidence: {item.get('evidence', '')}"
        )

    sections.append(
        "ACTION ITEMS\n" +
        "\n\n".join(actions)
    )

    unresolved = "\n".join(
        f"- {x}"
        for x in data.get(
            "unresolved_issues",
            []
        )
    )

    sections.append(
        f"UNRESOLVED ISSUES\n{unresolved}"
    )

    ambiguities = "\n".join(
        f"- {x}"
        for x in data.get(
            "ambiguities",
            []
        )
    )

    sections.append(
        f"AMBIGUITIES\n{ambiguities}"
    )

    return "\n\n".join(sections)


generic_readable = format_meeting_output(
    generic_run1_data
)

structured_readable = format_meeting_output(
    structured_run1_data
)


# ------------------------------------------------------------
# Randomized blinded assignment
# Different order for H1 and H2
# ------------------------------------------------------------

random.seed(42)

evaluator_mappings = {}

for evaluator_id in ["H1", "H2"]:

    conditions = [
        ("generic", generic_readable),
        ("structured", structured_readable)
    ]

    random.shuffle(conditions)

    evaluator_mappings[evaluator_id] = {
        "Output A": conditions[0][0],
        "Output B": conditions[1][0]
    }

    output_contents = {
        "Output A": conditions[0][1],
        "Output B": conditions[1][1]
    }


    # --------------------------------------------------------
    # Create workbook
    # --------------------------------------------------------

    wb = Workbook()

    ws_instruction = wb.active
    ws_instruction.title = "Instructions"

    instructions = [
        "M01 Blinded Human Evaluation",
        "",
        "Purpose:",
        "Evaluate two AI-generated meeting outputs without knowing which prompt produced them.",
        "",
        "Read the original meeting transcript first.",
        "Then evaluate Output A and Output B independently.",
        "",
        "Rating Scale:",
        "1 = Very Poor",
        "2 = Poor",
        "3 = Acceptable",
        "4 = Good",
        "5 = Excellent",
        "",
        "Evaluation Dimensions:",
        "Accuracy = factual correctness relative to the transcript.",
        "Completeness = coverage of important meeting information.",
        "Clarity = how clear and understandable the output is.",
        "Usefulness = practical usefulness as meeting documentation.",
        "",
        "Do not compare the outputs while assigning the first rating.",
        "Judge each output against the transcript.",
        "Do not attempt to identify which prompting method was used."
    ]

    for row_num, text in enumerate(
        instructions,
        start=1
    ):
        ws_instruction.cell(
            row=row_num,
            column=1,
            value=text
        )


    # --------------------------------------------------------
    # Transcript sheet
    # --------------------------------------------------------

    ws_transcript = wb.create_sheet(
        "Transcript"
    )

    ws_transcript["A1"] = (
        "Original Human-Reviewed Meeting Transcript"
    )

    ws_transcript["A1"].font = Font(
        bold=True
    )

    transcript_lines = (
        reviewed_transcript.splitlines()
    )

    for row_num, line in enumerate(
        transcript_lines,
        start=3
    ):
        ws_transcript.cell(
            row=row_num,
            column=1,
            value=line
        )


    # --------------------------------------------------------
    # Output A
    # --------------------------------------------------------

    ws_a = wb.create_sheet(
        "Output A"
    )

    ws_a["A1"] = "AI Meeting Output A"
    ws_a["A1"].font = Font(
        bold=True
    )

    ws_a["A3"] = output_contents[
        "Output A"
    ]

    ws_a["A3"].alignment = Alignment(
        wrap_text=True,
        vertical="top"
    )


    # --------------------------------------------------------
    # Output B
    # --------------------------------------------------------

    ws_b = wb.create_sheet(
        "Output B"
    )

    ws_b["A1"] = "AI Meeting Output B"
    ws_b["A1"].font = Font(
        bold=True
    )

    ws_b["A3"] = output_contents[
        "Output B"
    ]

    ws_b["A3"].alignment = Alignment(
        wrap_text=True,
        vertical="top"
    )


    # --------------------------------------------------------
    # Rating sheet
    # --------------------------------------------------------

    ws_rating = wb.create_sheet(
        "Ratings"
    )

    headers = [
        "meeting_id",
        "evaluator_id",
        "blind_output_id",
        "accuracy",
        "completeness",
        "clarity",
        "usefulness",
        "overall_comments"
    ]

    for col_num, header in enumerate(
        headers,
        start=1
    ):
        ws_rating.cell(
            row=1,
            column=col_num,
            value=header
        )

        ws_rating.cell(
            row=1,
            column=col_num
        ).font = Font(
            bold=True
        )


    for row_num, blind_id in enumerate(
        ["Output A", "Output B"],
        start=2
    ):

        values = [
            "M01",
            evaluator_id,
            blind_id,
            "",
            "",
            "",
            "",
            ""
        ]

        for col_num, value in enumerate(
            values,
            start=1
        ):
            ws_rating.cell(
                row=row_num,
                column=col_num,
                value=value
            )


    # Column widths
    for ws in wb.worksheets:
        ws.column_dimensions["A"].width = 45

    ws_rating.column_dimensions["H"].width = 50


    evaluation_file = (
        DIRS["human_results"] /
        f"M01_{evaluator_id}_Blinded_Human_Evaluation.xlsx"
    )

    wb.save(
        evaluation_file
    )

    print(
        f"{evaluator_id} workbook created:"
    )
    print(evaluation_file)


# ------------------------------------------------------------
# Save hidden condition mapping
# Evaluators should NOT receive this file
# ------------------------------------------------------------

mapping_file = (
    DIRS["logs"] /
    "M01_blinded_human_evaluation_mapping.json"
)

with open(
    mapping_file,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        evaluator_mappings,
        f,
        indent=2
    )


print("\nBlinded human evaluation package created successfully.")

print("\nEvaluators:")
print("- H1")
print("- H2")

print("\nEach evaluator will rate:")
print("- Output A")
print("- Output B")

print("\nDimensions:")
print("- Accuracy")
print("- Completeness")
print("- Clarity")
print("- Usefulness")

print("\nRating scale:")
print("1 = Very Poor")
print("2 = Poor")
print("3 = Acceptable")
print("4 = Good")
print("5 = Excellent")

print("\nCondition mapping saved privately:")
print(mapping_file)

print(
    "\nIMPORTANT: Do NOT give the mapping file to H1 or H2."
)

print(
    "The evaluators should receive only their respective Excel workbook."
)
```

```python
# ============================================================
# IMPORT COMPLETED BLINDED HUMAN EVALUATION
# H1 or H2
# ============================================================

import pandas as pd
from google.colab import files


def import_completed_human_evaluation(evaluator_id):
    """
    Upload and validate a completed blinded human-evaluation
    workbook for evaluator H1 or H2.
    """

    evaluator_id = evaluator_id.upper().strip()

    if evaluator_id not in ["H1", "H2"]:
        raise ValueError(
            "evaluator_id must be 'H1' or 'H2'."
        )

    print(
        f"Upload the COMPLETED workbook for {evaluator_id}."
    )

    uploaded = files.upload()

    if not uploaded:
        print("No file uploaded.")
        return None

    uploaded_filename = list(
        uploaded.keys()
    )[0]

    print("\nUploaded file:")
    print(uploaded_filename)


    # --------------------------------------------------------
    # Read Ratings sheet
    # --------------------------------------------------------

    try:

        ratings_df = pd.read_excel(
            uploaded_filename,
            sheet_name="Ratings"
        )

    except Exception as e:

        print(
            "\nERROR: Could not read the 'Ratings' sheet."
        )
        print(e)

        return None


    # --------------------------------------------------------
    # Required columns
    # --------------------------------------------------------

    required_columns = [
        "meeting_id",
        "evaluator_id",
        "blind_output_id",
        "accuracy",
        "completeness",
        "clarity",
        "usefulness",
        "overall_comments"
    ]

    missing_columns = [
        col
        for col in required_columns
        if col not in ratings_df.columns
    ]

    if missing_columns:

        print(
            "\nERROR: Missing required columns:"
        )

        print(missing_columns)

        return None


    # --------------------------------------------------------
    # Keep expected rows only
    # --------------------------------------------------------

    ratings_df = ratings_df[
        ratings_df["blind_output_id"].isin(
            ["Output A", "Output B"]
        )
    ].copy()


    if len(ratings_df) != 2:

        print(
            "\nERROR: Expected exactly two rating rows "
            "(Output A and Output B)."
        )

        return None


    # --------------------------------------------------------
    # Verify evaluator ID
    # --------------------------------------------------------

    ratings_df[
        "evaluator_id"
    ] = evaluator_id

    ratings_df[
        "meeting_id"
    ] = "M01"


    # --------------------------------------------------------
    # Validate rating values
    # --------------------------------------------------------

    rating_columns = [
        "accuracy",
        "completeness",
        "clarity",
        "usefulness"
    ]

    validation_errors = []

    for column in rating_columns:

        ratings_df[column] = pd.to_numeric(
            ratings_df[column],
            errors="coerce"
        )

        invalid_rows = ratings_df[
            ratings_df[column].isna()
            |
            (ratings_df[column] < 1)
            |
            (ratings_df[column] > 5)
        ]

        if not invalid_rows.empty:

            validation_errors.append(
                column
            )


    if validation_errors:

        print(
            "\nVALIDATION FAILED."
        )

        print(
            "The following dimensions contain "
            "blank or invalid ratings:"
        )

        for item in validation_errors:
            print("-", item)

        print(
            "\nEvery rating must be an integer "
            "from 1 to 5."
        )

        return None


    # Force integer values
    for column in rating_columns:

        ratings_df[column] = (
            ratings_df[column]
            .astype(int)
        )


    # --------------------------------------------------------
    # Save completed blinded ratings
    # --------------------------------------------------------

    output_file = (
        DIRS["human_results"] /
        f"M01_{evaluator_id}_completed_blinded_ratings.csv"
    )

    ratings_df.to_csv(
        output_file,
        index=False
    )


    print(
        "\nHuman evaluation imported successfully."
    )

    print("\nEvaluator:")
    print(evaluator_id)

    print("\nValidated ratings:")
    display(ratings_df)

    print("\nSaved file:")
    print(output_file)

    print(
        "\nIMPORTANT: Prompt identities remain blinded "
        "at this stage."
    )

    return ratings_df


print(
    "Completed human-evaluation importer created successfully."
)

print("\nUsage after evaluators return their files:")
print(
    "H1_ratings = import_completed_human_evaluation('H1')"
)
print(
    "H2_ratings = import_completed_human_evaluation('H2')"
)

print(
    "\nDo not run those two commands until the real "
    "evaluators have independently completed their workbooks."
)
```

```python
# ============================================================
# FORMAL INTER-ANNOTATOR AGREEMENT CALCULATOR
# Gold-standard annotators A1 vs A2
# ============================================================

import re
import numpy as np
import pandas as pd
from rapidfuzz.fuzz import token_set_ratio


# ------------------------------------------------------------
# Normalize free-text annotation
# ------------------------------------------------------------

def normalize_annotation_text(text):

    if pd.isna(text):
        return ""

    text = str(text).lower().strip()

    text = re.sub(
        r"[^\w\s]",
        " ",
        text
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


# ------------------------------------------------------------
# Pairwise item matching
# ------------------------------------------------------------

def calculate_annotation_agreement(
    annotator1_df,
    annotator2_df,
    similarity_threshold=70
):

    results = []

    for item_type in [
        "decision",
        "action_item"
    ]:

        a1 = annotator1_df[
            annotator1_df["item_type"] == item_type
        ].copy()

        a2 = annotator2_df[
            annotator2_df["item_type"] == item_type
        ].copy()

        a1 = a1[
            a1["item_text"].notna()
            &
            (
                a1["item_text"]
                .astype(str)
                .str.strip() != ""
            )
        ].reset_index(drop=True)

        a2 = a2[
            a2["item_text"].notna()
            &
            (
                a2["item_text"]
                .astype(str)
                .str.strip() != ""
            )
        ].reset_index(drop=True)


        # ----------------------------------------------------
        # Generate all possible pair similarities
        # ----------------------------------------------------

        candidate_pairs = []

        for i, row1 in a1.iterrows():

            for j, row2 in a2.iterrows():

                similarity = token_set_ratio(
                    normalize_annotation_text(
                        row1["item_text"]
                    ),
                    normalize_annotation_text(
                        row2["item_text"]
                    )
                )

                candidate_pairs.append({
                    "a1_index": i,
                    "a2_index": j,
                    "similarity": similarity
                })


        # Highest similarities matched first
        candidate_pairs = sorted(
            candidate_pairs,
            key=lambda x: x["similarity"],
            reverse=True
        )

        used_a1 = set()
        used_a2 = set()
        matched_pairs = []


        for pair in candidate_pairs:

            if (
                pair["similarity"]
                >= similarity_threshold
                and
                pair["a1_index"] not in used_a1
                and
                pair["a2_index"] not in used_a2
            ):

                used_a1.add(
                    pair["a1_index"]
                )

                used_a2.add(
                    pair["a2_index"]
                )

                matched_pairs.append(
                    pair
                )


        matched_count = len(
            matched_pairs
        )

        a1_count = len(a1)
        a2_count = len(a2)


        # ----------------------------------------------------
        # Agreement precision, recall and F1
        # ----------------------------------------------------

        agreement_precision = (
            matched_count / a1_count
            if a1_count > 0
            else 0.0
        )

        agreement_recall = (
            matched_count / a2_count
            if a2_count > 0
            else 0.0
        )

        if (
            agreement_precision
            +
            agreement_recall
        ) > 0:

            agreement_f1 = (
                2
                * agreement_precision
                * agreement_recall
                /
                (
                    agreement_precision
                    +
                    agreement_recall
                )
            )

        else:

            agreement_f1 = 0.0


        # ----------------------------------------------------
        # Mean semantic similarity of matched items
        # ----------------------------------------------------

        if matched_count > 0:

            mean_similarity = np.mean([
                pair["similarity"]
                for pair in matched_pairs
            ]) / 100

        else:

            mean_similarity = 0.0


        # ----------------------------------------------------
        # Owner/deadline agreement for action items
        # ----------------------------------------------------

        owner_agreements = []
        deadline_agreements = []

        if item_type == "action_item":

            for pair in matched_pairs:

                row1 = a1.iloc[
                    pair["a1_index"]
                ]

                row2 = a2.iloc[
                    pair["a2_index"]
                ]


                owner1 = normalize_annotation_text(
                    row1.get(
                        "owner",
                        ""
                    )
                )

                owner2 = normalize_annotation_text(
                    row2.get(
                        "owner",
                        ""
                    )
                )

                if owner1 or owner2:

                    owner_agreements.append(
                        int(owner1 == owner2)
                    )


                deadline1 = normalize_annotation_text(
                    row1.get(
                        "deadline",
                        ""
                    )
                )

                deadline2 = normalize_annotation_text(
                    row2.get(
                        "deadline",
                        ""
                    )
                )

                if deadline1 or deadline2:

                    deadline_agreements.append(
                        int(
                            deadline1
                            ==
                            deadline2
                        )
                    )


        owner_agreement = (
            np.mean(owner_agreements)
            if owner_agreements
            else None
        )

        deadline_agreement = (
            np.mean(deadline_agreements)
            if deadline_agreements
            else None
        )


        results.append({
            "meeting_id": "M01",
            "item_type": item_type,
            "annotator1_items": a1_count,
            "annotator2_items": a2_count,
            "matched_items": matched_count,
            "agreement_precision": round(
                agreement_precision,
                4
            ),
            "agreement_recall": round(
                agreement_recall,
                4
            ),
            "agreement_f1": round(
                agreement_f1,
                4
            ),
            "mean_matched_text_similarity": round(
                mean_similarity,
                4
            ),
            "owner_agreement": (
                round(
                    owner_agreement,
                    4
                )
                if owner_agreement
                is not None
                else None
            ),
            "deadline_agreement": (
                round(
                    deadline_agreement,
                    4
                )
                if deadline_agreement
                is not None
                else None
            ),
            "similarity_threshold": (
                similarity_threshold
            )
        })


    agreement_df = pd.DataFrame(
        results
    )

    return agreement_df


print(
    "Formal inter-annotator agreement calculator created successfully."
)

print("\nAgreement measures prepared:")
print("- Number of items identified by A1")
print("- Number of items identified by A2")
print("- One-to-one matched items")
print("- Agreement precision")
print("- Agreement recall")
print("- Agreement F1")
print("- Mean matched-text similarity")
print("- Owner agreement")
print("- Deadline agreement")

print("\nDefault matching threshold:")
print("70/100 token-set similarity")

print(
    "\nIMPORTANT: No agreement score has been calculated yet."
)

print(
    "Real completed A1 and A2 annotations are required before "
    "this function can be executed."
)

print(
    "\nThis free-text extraction agreement should be reported "
    "as pairwise annotation agreement, not mislabelled as Cohen's kappa."
)
```

```python
# ============================================================
# M01 EXTRACTION PERFORMANCE EVALUATION
# Generic vs Structured Prompt
# ============================================================

import json
import pandas as pd
from rapidfuzz.fuzz import token_set_ratio


# ------------------------------------------------------------
# Load gold standard
# ------------------------------------------------------------

gold_file = (
    DIRS["gold_standard"] /
    "M01_consensus_gold_standard.csv"
)

gold_df = pd.read_csv(
    gold_file
)


# ------------------------------------------------------------
# Load Gemini outputs
# ------------------------------------------------------------

def load_output(prompt_type):

    output_file = (
        DIRS[f"{prompt_type}_outputs"] /
        f"M01_{prompt_type}_run1.json"
    )

    with open(
        output_file,
        "r",
        encoding="utf-8"
    ) as f:
        return json.load(f)



# ------------------------------------------------------------
# Extract AI items
# ------------------------------------------------------------

def extract_ai_items(output):

    items = []

    for item in output.get(
        "decisions",
        []
    ):

        items.append({
            "item_type": "decision",
            "item_text": item["decision"],
            "owner": "",
            "deadline": ""
        })


    for item in output.get(
        "action_items",
        []
    ):

        items.append({
            "item_type": "action_item",
            "item_text": item["action"],
            "owner": item.get(
                "owner",
                ""
            ),
            "deadline": item.get(
                "deadline",
                ""
            )
        })

    return pd.DataFrame(items)



# ------------------------------------------------------------
# Matching function
# ------------------------------------------------------------

def evaluate_extraction(
    ai_df,
    gold_df,
    threshold=70
):

    matched_gold = set()
    matched_ai = set()

    matches = []


    for ai_index, ai_row in ai_df.iterrows():

        best_match = None
        best_score = 0


        for gold_index, gold_row in gold_df.iterrows():

            if gold_index in matched_gold:
                continue


            if (
                ai_row["item_type"]
                !=
                gold_row["item_type"]
            ):
                continue


            score = token_set_ratio(
                str(ai_row["item_text"]),
                str(gold_row["item_text"])
            )


            if score > best_score:

                best_score = score
                best_match = gold_index


        if (
            best_match is not None
            and best_score >= threshold
        ):

            matched_ai.add(ai_index)
            matched_gold.add(best_match)

            matches.append({
                "ai_index": ai_index,
                "gold_index": best_match,
                "similarity": best_score
            })


    TP = len(matches)

    FP = len(ai_df) - TP

    FN = len(gold_df) - len(matched_gold)


    precision = (
        TP / (TP + FP)
        if (TP + FP) > 0
        else 0
    )

    recall = (
        TP / (TP + FN)
        if (TP + FN) > 0
        else 0
    )

    f1 = (
        2 * precision * recall /
        (precision + recall)
        if precision + recall > 0
        else 0
    )


    hallucination_rate = (
        FP / len(ai_df)
        if len(ai_df) > 0
        else 0
    )


    # ---------------------------------------------
    # Owner and deadline accuracy
    # ---------------------------------------------

    owner_correct = []
    deadline_correct = []


    for match in matches:

        ai_item = ai_df.iloc[
            match["ai_index"]
        ]

        gold_item = gold_df.iloc[
            match["gold_index"]
        ]


        if gold_item["item_type"] == "action_item":

            owner_correct.append(
                int(
                    str(ai_item["owner"]).lower().strip()
                    ==
                    str(gold_item["owner"]).lower().strip()
                )
            )

            deadline_correct.append(
                int(
                    str(ai_item["deadline"]).lower().strip()
                    ==
                    str(gold_item["deadline"]).lower().strip()
                )
            )


    return {
        "TP": TP,
        "FP": FP,
        "FN": FN,
        "precision": round(precision,4),
        "recall": round(recall,4),
        "f1_score": round(f1,4),
        "hallucination_rate": round(
            hallucination_rate,
            4
        ),
        "owner_accuracy": round(
            sum(owner_correct)/len(owner_correct),
            4
        ) if owner_correct else None,
        "deadline_accuracy": round(
            sum(deadline_correct)/len(deadline_correct),
            4
        ) if deadline_correct else None
    }



# ------------------------------------------------------------
# Run evaluation
# ------------------------------------------------------------

results = []


for prompt_type in [
    "generic",
    "structured"
]:

    output = load_output(
        prompt_type
    )

    ai_df = extract_ai_items(
        output
    )

    metrics = evaluate_extraction(
        ai_df,
        gold_df
    )

    metrics["prompt_type"] = prompt_type

    results.append(
        metrics
    )


results_df = pd.DataFrame(
    results
)


# Arrange columns

results_df = results_df[
    [
        "prompt_type",
        "TP",
        "FP",
        "FN",
        "precision",
        "recall",
        "f1_score",
        "hallucination_rate",
        "owner_accuracy",
        "deadline_accuracy"
    ]
]


print(
    "M01 extraction evaluation completed successfully."
)

display(
    results_df
)


results_file = (
    DIRS["results"] /
    "M01_generic_vs_structured_extraction_metrics.csv"
)


results_df.to_csv(
    results_file,
    index=False
)


print("\nSaved file:")
print(results_file)

print(
    "\nInterpretation:"
)

print(
    "Metrics represent agreement with the M01 controlled gold standard."
)

print(
    "This is a prototype case-study evaluation and should be reported with limitations."
)
```
